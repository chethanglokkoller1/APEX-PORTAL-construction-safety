## 3. CORE MODULES OVERVIEW

### 3.1 Detection Module (`src/detection.py`)

**Purpose:** Real-time YOLOv8-based safety violation detection

**Key Features:**
- Multi-format image input support (file path, numpy array, PIL Image)
- PPE detection (helmet, safety vest)
- Worker detection
- Equipment detection (excavator, crane, truck, bulldozer)
- Real-time inference
- Confidence-based filtering

**Main Classes:**
- `SafetyDetector`: Primary detection engine
- Methods: `detect_image()`, `detect_frame()`, `detect_video()`, `process_detections()`, `draw_detections()`

**Key Methods:**

```python
detector = SafetyDetector(
    model_path='models/safety_detection/weights/best.pt',
    conf_threshold=0.5,
    iou_threshold=0.4,
    device='cpu'  # or 'cuda' for GPU
)

# Single image detection
result = detector.detect_image('path/to/image.jpg')
# Returns: {'image': annotated_array, 'detections': [...]}

# Video inference
detector.detect_video('path/to/video.mp4', output_path='output.mp4')

# Webcam real-time detection
detector.webcam_inference()
```

- Unique worker ID assignment

- Historical trajectory storage
- Zone-based tracking

**Main Classes:**
- `ObjectTracker`: ByteTrack-based tracker

**Usage:**

```python
tracker = ObjectTracker()

# Update tracker with detections
tracked_objects = tracker.update(detections, frame)
# Returns: List of tracked objects with IDs and trajectories

# Get zone violations
zone_violations = tracker.check_zone_violations(zone_polygon, tracked_objects)
```

### 3.3 Analytics Module (`src/analytics.py`)

**Purpose:** Safety compliance metrics and risk analysis

**Key Features:**
- Real-time violation logging
- Compliance score calculation (0-100)
- Risk level assessment (High/Medium/Low)
- Trend analysis (7-day, 30-day)
- Safety metrics dashboard
- Incident summarization

**Database Schema:**

```sql
-- Violations Table
CREATE TABLE violations (
    id INTEGER PRIMARY KEY,
    timestamp DATETIME,
    violation_type TEXT,  -- 'no_helmet', 'no_vest', 'unauthorized_zone', etc.
    severity TEXT,        -- 'high', 'medium', 'low'
    confidence REAL,
    worker_id TEXT,
    zone TEXT
);

-- Compliance Table
CREATE TABLE compliance (
    id INTEGER PRIMARY KEY,
    date DATE,
    compliance_score REAL,  -- 0-100
    total_workers INTEGER,
    violations_count INTEGER
);
```

**Key Methods:**

```python
analytics = SafetyAnalytics(db_path='reports/safety_analytics.db')

# Log a violation
analytics.log_violation(
    violation_type='no_helmet',
    severity='high',
    confidence=0.95,
    worker_id='W001'
)

# Get compliance metrics
compliance_trend = analytics.get_compliance_trend(days=30)
safety_score = analytics.get_dashboard_metrics()
recent_violations = analytics.get_recent_violations(limit=10)

# Risk analysis
risk_trend = analytics.get_risk_trend(days=30)
incident_summary = analytics.get_incident_summary()
```

## 8. USAGE EXAMPLES

### Example 1: Detect in a Single Image

```python
from src.detection import SafetyDetector
from PIL import Image
import cv2

detector = SafetyDetector()

# Method 1: File path
result = detector.detect_image('data/images/site_01.jpg')

# Method 2: PIL Image
pil_image = Image.open('data/images/site_01.jpg')
result = detector.detect_image(pil_image)

# Method 3: NumPy array
cv_image = cv2.imread('data/images/site_01.jpg')
result = detector.detect_image(cv_image)

# Process results
annotated_image = result['image']
detections = result['detections']

for detection in detections:
    print(f"Class: {detection['class']}, Confidence: {detection['confidence']:.2%}")

# Display
cv2.imshow('Detections', annotated_image)
cv2.waitKey(0)
```

### Example 2: Process Video with Tracking

```python
from src.detection import SafetyDetector
from src.tracking import ObjectTracker
from src.analytics import SafetyAnalytics
import cv2

detector = SafetyDetector()
tracker = ObjectTracker()
analytics = SafetyAnalytics()

cap = cv2.VideoCapture('data/videos/site_footage.mp4')

while True:
    ret, frame = cap.read()
    if not ret:
        break
    
    # Detect
    detections = detector.detect_frame(frame)
    processed = detector.process_detections(detections)
    
    # Track
    tracked = tracker.update(processed, frame)
    
    # Log violations
    for det in processed:
        if det['class'] == 'no_helmet':
            analytics.log_violation(
                violation_type='no_helmet',
                severity='high',
                confidence=det['confidence']
            )
    
    # Annotate
    annotated = detector.draw_detections(frame, processed)
    cv2.imshow('Live Detection', annotated)
    
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

### Example 3: Generate Safety Report

```python
from src.analytics import SafetyAnalytics
import json
from datetime import datetime

analytics = SafetyAnalytics()

# Collect metrics
report = {
    'generated_at': datetime.now().isoformat(),
    'safety_score': analytics.get_dashboard_metrics()['safety_score'],
    'compliance_trend': analytics.get_compliance_trend(days=30).to_dict(),
    'recent_violations': analytics.get_recent_violations(limit=10),
    'risk_assessment': analytics.get_risk_trend(days=30).to_dict(),
    'incident_summary': analytics.get_incident_summary(days=30)
}

# Export
with open('reports/safety_report.json', 'w') as f:
    json.dump(report, f, indent=2, default=str)

print("Report saved to reports/safety_report.json")
```

## 10. TROUBLESHOOTING & FAQ

### Common Issues

**Issue 1: "Model not found" warning**
- **Cause**: Custom model `best.pt` not present
- **Solution**: System automatically falls back to `yolov8n.pt` (fine-tuned YOLOv8 nano)
- **Status**: ✅ Expected behavior

**Issue 2: "AttributeError: 'numpy.ndarray' has no attribute 'flush'"**
- **Cause**: YOLO treating numpy array as file path
- **Solution**: Already fixed! Numpy arrays wrapped in list before passing to model
- **Status**: ✅ RESOLVED (June 10, 2026)

**Issue 3: Streamlit warnings about deprecated parameters**
- **Cause**: `use_container_width` deprecated in newer Streamlit versions
- **Solution**: Non-critical warnings. Dashboard functions normally
- **Status**: ⚠️ Future update needed (minor)

**Issue 4: Database locked error**
- **Cause**: Multiple processes accessing SQLite simultaneously
- **Solution**: Stop all Python processes, restart dashboard
- **Command**: `taskkill /F /IM python.exe`

**Issue 5: Out of Memory (OOM)**
- **Cause**: Large video files or long inference sessions
- **Solution**: Reduce frame skip rate, use smaller resolution
- **Code**: Set `frame_skip=5` in `detect_video()`

### Performance Optimization

```python
# Lower confidence threshold = faster but more false positives
detector = SafetyDetector(conf_threshold=0.3)

# Skip frames in video processing
detector.detect_video('video.mp4', frame_skip=3)

# Use GPU if available
detector = SafetyDetector(device='cuda')
```
